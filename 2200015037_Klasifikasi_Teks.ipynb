##Membaca Dataset

```python
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/mohitgupta-omg/Kaggle-SMS-Spam-Collection-Dataset-/master/spam.csv',
encoding='latin-1')
data.head()
```

```python
# menghapus kolom yang tidak dikenal dan mengubah nama kolom
data.drop(['Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'], axis=1, inplace=True)
data.columns = ['label', 'text']
data.head()
```

##Exploratory Data Analysis (EDA)

```python
# check missing values
data.isna().sum()
```

```python
# check data shape
data.shape
```

```python
# check target balance
data['label'].value_counts(normalize = True).plot.bar()
```

##Text Preprocessing

```python
# download nltk

import nltk
nltk.download(['stopwords','wordnet','omw-1.4'])
# create a list text
text = list(data['text'])
# preprocessing loop
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
corpus = []
for i in range(len(text)):
  r = re.sub('[^a-zA-Z]', ' ', text[i])
  r = r.lower()
  r = r.split()
  r = [word for word in r if word not in stopwords.words('english')]
  r = [lemmatizer.lemmatize(word) for word in r]
  r = ' '.join(r)
  corpus.append(r)
#assign corpus to data['text']
data['text'] = corpus
data.head()
```

## Pemisahan Train - Test

```python
# Create Feature and Label sets
X = data['text']
y = data['label']
# train test split (66% train - 33% test)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=123)
print('Training Data :', X_train.shape)
print('Testing Data : ', X_test.shape)
```

## Ekstraksi Fitur

```python
# Train Bag of Words model
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X_train_cv = cv.fit_transform(X_train)
X_train_cv.shape
```

##Training

```python
# Training Logistic Regression model
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train_cv, y_train)
```

##Prediksi

```python
# transform X_test using CV
X_test_cv = cv.transform(X_test)
# generate predictions
predictions = lr.predict(X_test_cv)
predictions
```

## Evaluasi

```python
# confusion matrix
import pandas as pd
from sklearn import metrics
df = pd.DataFrame(metrics.confusion_matrix(y_test,predictions), index=['ham','spam'], columns=['ham','spam'])
df
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
display = ConfusionMatrixDisplay(metrics.confusion_matrix(y_test,predictions),display_labels=lr.classes_)
display.plot()
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test,predictions)
print('Accuracy :', accuracy)
```
